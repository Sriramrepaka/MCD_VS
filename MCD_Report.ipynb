## Filter Specifications

The filter consists of four distinct topology types, each realized using operational amplifiers (OpAmps):
* **Band-Pass (BP)**
* **Low-Pass (LP)**
* **High-Pass (HP)**
* **Band-Stop (BS)**

In this project, a fully integrated second-order biquad filter is designed, whose key specifications are a center frequency of $f_0 = 1 \text{ kHz}$, a quality factor of $Q = 10$, and a DC pass-band gain of $H_0 = 1$. 

| Parameter | Symbol | Target Value | 
|---|---|---|
| **Center Frequency**(Center frequency of the narrowband response)  | $f_0$ | -- |
| **Quality Factor**(Determines selectivity and sharpness) | $Q$ | -- |
| **DC Pass-Band Gain**(Peak gain of unity at low frequencies) | $H_0$ | -- |

These parameters ensure a narrowband response around 1 kHz with a peak gain of unity at low frequencies and sharp roll-off in the stop-bands.
